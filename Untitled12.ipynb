```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras import layers, Model
```

```python
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
print("Original Training Shape:", x_train.shape)
print("Original Testing Shape :", x_test.shape)
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)
print("Training Shape:", x_train.shape)
print("Testing Shape :", x_test.shape)
```

```python
latent_dim = 2
input_shape = (28, 28, 1)
```

```python
encoder_inputs = layers.Input(
 shape=input_shape,
 name="encoder_input"
)
x = layers.Flatten()(encoder_inputs)
x = layers.Dense(
 256,
 activation="relu"
)(x)
x = layers.Dense(
 128,
 activation="relu"
)(x)
```

```python
z_mean = layers.Dense(
 latent_dim,
 name="z_mean"
)(x)
```

```python
z_log_var = layers.Dense(
 latent_dim,
 name="z_log_var"
)(x)
```

```python
class Sampling(layers.Layer):
 def call(self, inputs):
  z_mean, z_log_var = inputs
  # Random noise
  epsilon = tf.random.normal(
  shape=tf.shape(z_mean)
  )
  # Reparameterization trick
  z = (
  z_mean
  + tf.exp(0.5 * z_log_var)
  * epsilon
  )
  return z
  z = Sampling(name="sampling")(
  [z_mean, z_log_var]
  )
```

```python
 encoder = Model(
 encoder_inputs,
 [z_mean, z_log_var, z],
 name="encoder"
)
print("\n==============================")
print("ENCODER SUMMARY")
print("==============================")
encoder.summary()
```

```python
latent_inputs = layers.Input(
 shape=(latent_dim,),
 name="decoder_input"
)
x = layers.Dense(
 128,
 activation="relu"
)(latent_inputs)
x = layers.Dense(
 256,
 activation="relu"
)(x)
x = layers.Dense(
 28 * 28,
 activation="sigmoid"
)(x)
decoder_outputs = layers.Reshape(
 (28, 28, 1)
)(x)
```

```python
oder = Model(
 latent_inputs,
 decoder_outputs,
 name="decoder"
)
print("\n==============================")
print("DECODER SUMMARY")
print("==============================")
oder.summary()
```

```python
class VAE(Model):
 def __init__(
 self,
 encoder,
 decoder,
 **kwargs
 ):
  super().__init__(**kwargs)
  self.encoder = encoder
  self.decoder = decoder
  # ----------------------------------------------------
  # Metrics
  # ----------------------------------------------------
  self.total_loss_tracker = tf.keras.metrics.Mean(
  name="total_loss"
  )
  self.reconstruction_loss_tracker = tf.keras.metrics.Mean(
  name="reconstruction_loss"
  )
  self.kl_loss_tracker = tf.keras.metrics.Mean(
  name="kl_loss"
  )
```

```python
@property
def metrics(self):
 return [
 self.total_loss_tracker,
 self.reconstruction_loss_tracker,
 self.kl_loss_tracker
 ]
```

```python
def call(
 self,
 inputs,
 training=False
 ):
  z_mean, z_log_var, z = self.encoder(
  inputs,
  training=training
  )
  # Decoder
  reconstruction = self.decoder(
  z,
  training=training
  )
  return reconstruction
```
